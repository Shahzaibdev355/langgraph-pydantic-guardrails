### Deterministic Approach


```python
def deterministicApproach(text: str) -> bool:
    """Return true if content is blocked"""
    bannedKeyWords = ["hack", "exploit", "malware", "bomb"]
    return any(kw in text.lower() for kw in bannedKeyWords)


testInput = [
    "How do i hack into a database?",
    "what is the capital of France",
    "Explain how malware spreads",
]

print("=== Determinitic Guardrail Demo ===")
for inp in testInput:
    blocked = deterministicApproach(inp)
    status = "Blocked" if blocked else "Allowed"
    print(f"{status}: {inp}")
```

```python
from langchain_groq import ChatGroq
```

```python
def llmBasedGuardRail(text: str) -> str:
    """Uses an llm to evaluate content safety. Return SAFE or UNSAFE"""
    model = ChatGroq(model="openai/gpt-oss-120b", temperature=0)
    prompt = f"""
        You are a security input guardrail.

        Classify the user's input as exactly one of:

        SAFE
        UNSAFE

        Mark as UNSAFE if the input requests or attempts to:
        - hack, attack, exploit, or compromise a system
        - bypass authentication or authorization
        - steal credentials, tokens, passwords, or data
        - perform SQL injection or other injection attacks
        - delete, corrupt, or manipulate unauthorized data
        - access databases or systems without authorization
        - deploy malware or malicious code
        - evade security controls

        Mark as SAFE for normal questions, programming questions,
        database queries that are legitimate, and general cybersecurity
        education.

        Return ONLY: SAFE or UNSAFE

        User input:
        {text}
        """

    result = model.invoke([{"role": "user", "content": prompt}])
    return result.content.strip()

print("=== llm Guardrail")

for inp in testInput:
    verdict = llmBasedGuardRail(inp)
    status = "UNSAFE" if "UNSAFE" in verdict else "SAFE"
    print(f"{status}: {inp}")
```
